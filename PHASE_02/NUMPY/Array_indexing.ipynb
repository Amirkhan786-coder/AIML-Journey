NUMPY INDEXING AND SLICING OF VECTORS

```python
arr = np.arange(11,21)
arr
```

```python
arr[8]
```

```python
arr_slice = arr[2:8:2]
arr_slice
```

NUMPY INDEXING AND SLICING OF MATRIX

```python
arr = np.arange(1,31).reshape(6,5)
arr
```

```python
arr[0]
```

```python
arr[3,3]
```

```python
arr[0,4]
```

```python
slice = arr[3:,3:]
slice
```
