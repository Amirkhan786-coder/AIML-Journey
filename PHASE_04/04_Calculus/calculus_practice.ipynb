27.4 Python Implementation

```python
# Define learning rate
learning_rate = 0.1

# Initial parameter
w = 0.0

# Number of iterations
epochs = 20

for epoch in range(epochs):

    # Calculate gradient
    gradient = 2 * (w - 3)

    # Update parameter
    w = w - learning_rate * gradient

    # Calculate loss
    loss = (w - 3) ** 2

    print(
        f"Epoch: {epoch + 1}, "
        f"Weight: {w:.4f}, "
        f"Loss: {loss:.4f}"
    )

print("Final Weight:", w)
```

32.2 Partial Derivatives Using SymPy

```python
import sympy as sp

x, y = sp.symbols('x y')

f = x**2 + 3*x*y + y**2

df_dx = sp.diff(f, x)
df_dy = sp.diff(f, y)

print("Partial derivative w.r.t x:", df_dx)
print("Partial derivative w.r.t y:", df_dy)
```
