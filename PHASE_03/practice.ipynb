```python
def missing_value_report(data):

    # Count missing values in each column
    missing_count = data.isnull().sum()

    # Calculate missing percentage
    missing_percent = (
        missing_count / len(data)
    ) * 100

    # Create a report DataFrame
    report = pd.DataFrame({
        "Missing_Count": missing_count,
        "Missing_Percentage": missing_percent
    })

    # Return the report
    return report


# Call the function
report = missing_value_report(df)

print(report)
```

```python
def detect_outliers_iqr(data, column):

    # Calculate Q1
    Q1 = data[column].quantile(0.25)

    # Calculate Q3
    Q3 = data[column].quantile(0.75)

    # Calculate IQR
    IQR = Q3 - Q1

    # Calculate boundaries
    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    # Filter outliers
    outliers = data[
        (data[column] < lower_limit) |
        (data[column] > upper_limit)
    ]

    return outliers


# Detect Marks outliers
marks_outliers = detect_outliers_iqr(df, "Marks")

# Detect Study Hours outliers
hours_outliers = detect_outliers_iqr(
    df,
    "Study_Hours_Per_Week"
)

print("Marks Outliers:")
print(marks_outliers)

print("Study Hours Outliers:")

print(hours_outliers)
```

```python
def clean_student_data(data):

    # Create a copy
    cleaned = data.copy()

    # Remove exact duplicate rows
    cleaned = cleaned.drop_duplicates()

    # Standardize City names
    cleaned["City"] = (
        cleaned["City"]
        .astype("string")
        .str.strip()
        .str.title()
    )

    # Convert invalid Age to missing
    cleaned.loc[
        (cleaned["Age"] < 15) |
        (cleaned["Age"] > 100),
        "Age"
    ] = np.nan

    # Convert invalid Attendance to missing
    cleaned.loc[
        (cleaned["Attendance"] < 0) |
        (cleaned["Attendance"] > 100),
        "Attendance"
    ] = np.nan

    # Convert invalid Marks to missing
    cleaned.loc[
        (cleaned["Marks"] < 0) |
        (cleaned["Marks"] > 100),
        "Marks"
    ] = np.nan

    # Convert negative Study Hours to missing
    cleaned.loc[
        cleaned["Study_Hours_Per_Week"] < 0,
        "Study_Hours_Per_Week"
    ] = np.nan

    # Numerical columns
    numerical_columns = [
        "Age",
        "Study_Hours_Per_Week",
        "Attendance",
        "Marks"
    ]

    # Fill numerical missing values using median
    for column in numerical_columns:

        median_value = cleaned[column].median()

        cleaned[column] = cleaned[column].fillna(
            median_value
        )

    # Fill categorical missing values using mode
    categorical_columns = ["Name", "City"]

    for column in categorical_columns:

        mode_value = cleaned[column].mode()

        if not mode_value.empty:

            cleaned[column] = cleaned[column].fillna(
                mode_value[0]
            )

    return cleaned


# Run cleaning function
df_cleaned = clean_student_data(df)

print("Original Shape:", df.shape)

print("Cleaned Shape:", df_cleaned.shape)

print("\nMissing Values After Cleaning:")

print(df_cleaned.isnull().sum())

print("\nCleaned Dataset:")

print(df_cleaned.head())
```

```python
def transform_student_data(data):

    # Create a copy
    transformed = data.copy()

    # Create Performance column
    conditions = [
        transformed["Marks"] >= 90,
        transformed["Marks"] >= 75,
        transformed["Marks"] >= 60
    ]

    categories = [
        "Excellent",
        "Good",
        "Average"
    ]

    transformed["Performance"] = np.select(
        conditions,
        categories,
        default="Needs Improvement"
    )

    # Create Pass/Fail column
    transformed["Result"] = np.where(
        transformed["Marks"] >= 40,
        "Pass",
        "Fail"
    )

    # Scale numerical columns
    numerical_columns = [
        "Age",
        "Study_Hours_Per_Week",
        "Attendance",
        "Marks"
    ]

    scaler = MinMaxScaler()

    transformed[numerical_columns] = (
        scaler.fit_transform(
            transformed[numerical_columns]
        )
    )

    return transformed


# Run transformation
df_transformed = transform_student_data(df_cleaned)

print(df_transformed.head())
```

```python
def generate_insights(data):

    print("========== STUDENT PERFORMANCE INSIGHTS ==========")

    # 1. Total students
    print("\n1. Total Students:")

    print(len(data))

    # 2. Average marks
    print("\n2. Average Marks:")

    print(data["Marks"].mean())

    # 3. Highest marks
    print("\n3. Highest Marks:")

    print(data["Marks"].max())

    # 4. Lowest marks
    print("\n4. Lowest Marks:")

    print(data["Marks"].min())

    # 5. Average attendance
    print("\n5. Average Attendance:")

    print(data["Attendance"].mean())

    # 6. Average study hours
    print("\n6. Average Study Hours:")

    print(data["Study_Hours_Per_Week"].mean())

    # 7. Highest scoring student
    print("\n7. Highest Scoring Student:")

    print(
        data.loc[
            data["Marks"].idxmax(),
            ["Name", "Marks"]
        ]
    )

    # 8. City with highest average marks
    print("\n8. Best Performing City:")

    print(
        data.groupby("City")["Marks"].mean().idxmax()
    )

    # 9. Study hours and marks correlation
    print("\n9. Study Hours vs Marks Correlation:")

    print(
        data["Study_Hours_Per_Week"].corr(
            data["Marks"]
        )
    )

    # 10. Students with marks above 80
    print("\n10. Students Scoring Above 80:")

    print((data["Marks"] > 80).sum())


# Generate insights from cleaned data
generate_insights(df_cleaned)
```
